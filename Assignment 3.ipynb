```python

import random

machines = ['A', 'B', 'C']

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

epsilon = 0.2
N_play = 10
total_reward = 0

print("MODIFIED EPSILON-GREEDY MULTI-ARMED BANDIT")
print("------------------------------------------")

for play in range(1, N_play + 1):

    r = random.random()

    max_q = max(Q.values())

    best_machines = [
        m for m in machines if Q[m] == max_q
    ]

    best_machine = random.choice(best_machines)

    if r < epsilon:
        other_machines = [
            m for m in machines if m != best_machine
        ]
        selected = random.choice(other_machines)
        print("\nExploration")
    else:
        selected = best_machine
        print("\nExploitation")

    print("Play Number:", play)
    print("Random Number:", round(r, 3))
    print("Selected Machine:", selected)

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", {
        m: round(Q[m], 2) for m in machines
    })

print("\n========== FINAL RESULTS ==========")

print("\nMachine\tCount(N)\tReward(R)\tQ-Value")

for m in machines:
    print(
        m, "\t",
        N[m], "\t\t",
        R[m], "\t\t",
        round(Q[m], 2)
    )

print("\nTotal Reward Received:", total_reward)
print("Best Machine:", max(Q, key=Q.get))

print("===================================")
```

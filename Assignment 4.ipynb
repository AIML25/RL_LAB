```python

import math

machines = ['A', 'B', 'C']

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

N_play = 10
c = 2
total_reward = 0

print("UPPER CONFIDENCE BOUND - MULTI ARMED BANDIT")
print("-------------------------------------------")

for play in range(1, N_play + 1):

    if play <= 3:
        selected = machines[play - 1]

    else:
        t = play - 1

        for m in machines:
            UCB[m] = Q[m] + c * math.sqrt(
                math.log(t) / N[m]
            )

        selected = max(UCB, key=UCB.get)

    print("\nPlay Number:", play)
    print("Selected Machine:", selected)

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("N Table:", N)
    print("R Table:", R)
    print("Q Table:", {
        m: round(Q[m], 2) for m in machines
    })

    if play >= 3:
        print("UCB Table:", {
            m: round(UCB[m], 2) for m in machines
        })

for m in machines:
    UCB[m] = Q[m] + c * math.sqrt(
        math.log(N_play) / N[m]
    )

print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\tQ")

for m in machines:
    print(
        m, "\t",
        N[m], "\t",
        R[m], "\t",
        round(UCB[m], 2), "\t",
        round(Q[m], 2)
    )

print("\nTotal Reward Received:", total_reward)
print("Best Machine:", max(Q, key=Q.get))

print("===================================")
```

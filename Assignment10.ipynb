```python

gamma = 0.9
theta = 0.000001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']


transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },

    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 10)
    }
}


print("=" * 55)
print("DETERMINISTIC GRID WORLD")
print("=" * 55)

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Cell Reward: -1")
print("Wall Reward: -1")
print("Goal Reward: +10")


print("\nTRANSITION DETAILS")
print("-" * 55)

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]

        print(
            f"State: {state:2} | "
            f"Action: {action:5} | "
            f"Next State: {next_state:2} | "
            f"Reward: {reward}"
        )


V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

print("\nINITIAL STATE VALUES")
print(V)


print("\nVALUE ITERATION")
print("-" * 55)

iteration = 0

print(f"{'Iteration':<12}{'V(S0)':<15}"
      f"{'V(S1)':<15}{'Delta':<15}")

while True:


    new_V = V.copy()

    for state in states:

        q_values = []

        for action in actions:

            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]

            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(
        abs(new_V[state] - V[state])
        for state in states
    )

    iteration += 1
    V = new_V

    print(
        f"{iteration:<12}"
        f"{V['S0']:<15.4f}"
        f"{V['S1']:<15.4f}"
        f"{delta:<15.6f}"
    )


    if delta < theta:
        break


print("\nFINAL OPTIMAL STATE VALUES")
print("-" * 55)

for state in ['S0', 'S1', 'G']:
    print(f"V*({state}) = {V[state]:.4f}")

Q = {}

print("\nACTION-VALUE FUNCTION Q*(s, a)")
print("-" * 55)

for state in states:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        Q[state][action] = (
            reward + gamma * V[next_state]
        )

        print(
            f"Q*({state}, {action}) "
            f"= {Q[state][action]:.4f}"
        )

print("\nACTION-VALUE TABLE")
print("-" * 65)

print(f"{'State':<10}", end="")

for action in actions:
    print(f"{action:<12}", end="")

print()

for state in states:

    print(f"{state:<10}", end="")

    for action in actions:
        print(f"{Q[state][action]:<12.4f}", end="")

    print()


print("\nOPTIMAL POLICY")
print("-" * 55)

for state in states:

    maximum_q = max(Q[state].values())

    best_actions = [
        action
        for action in actions
        if abs(Q[state][action] - maximum_q) < 1e-9
    ]

    print(f"State {state}: {best_actions}")

print("\nBELLMAN OPTIMALITY VERIFICATION")
print("-" * 55)

for state in states:

    max_q = max(Q[state].values())

    verified = abs(V[state] - max_q) < 1e-9

    print(f"State: {state}")
    print(f"V*({state}) = {V[state]:.4f}")
    print(f"Maximum Q-value = {max_q:.4f}")
    print(f"Verified: {verified}")
    print()

print("RESULT: Value Iteration completed successfully.")
```

```python

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.000001

positions = {
    'S0': (0, 0),
    'S1': (1, 0),
    'G': (1, 1)
}

wall = (0, 1)

moves = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

def transition(state, action):
    if state == 'G':
        return 'G', 0

    r, c = positions[state]
    dr, dc = moves[action]
    new_position = (r + dr, c + dc)

    if (new_position == wall or
        new_position[0] < 0 or new_position[0] > 1 or
        new_position[1] < 0 or new_position[1] > 1):
        return state, -1

    if new_position == positions['G']:
        return 'G', 100

    for s, pos in positions.items():
        if pos == new_position:
            return s, -1

policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("TRANSITION INFORMATION")
print("----------------------")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transition(state, action)
        print(state, action, "->", next_state, "Reward:", reward)

def policy_evaluation(policy, V):
    while True:
        new_V = V.copy()

        for state in ['S0', 'S1']:
            action = policy[state]
            next_state, reward = transition(state, action)
            new_V[state] = reward + gamma * V[next_state]

        delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])
        V = new_V

        if delta < theta:
            break

    return V

def calculate_Q(state, action, V):
    next_state, reward = transition(state, action)
    return reward + gamma * V[next_state]

def policy_improvement(policy, V):
    new_policy = {}

    for state in ['S0', 'S1']:
        q_values = {
            action: calculate_Q(state, action, V)
            for action in actions
        }

        new_policy[state] = max(q_values, key=q_values.get)

    return new_policy

iteration = 1

while True:
    print("\nITERATION", iteration)

    V = policy_evaluation(policy, V)

    print("State Values:")
    for state in states:
        print(state, ":", round(V[state], 4))

    new_policy = policy_improvement(policy, V)

    print("Old Policy:", policy)
    print("New Policy:", new_policy)

    if new_policy == policy:
        print("Policy is Stable")
        break

    policy = new_policy
    iteration += 1

print("\nFINAL OPTIMAL STATE VALUES")
for state in states:
    print(state, ":", round(V[state], 4))

print("\nFINAL ACTION VALUE TABLE")

print("{:<10} {:<12} {:<12}".format("State", "Action", "Q Value"))

for state in ['S0', 'S1']:
    for action in actions:
        q = calculate_Q(state, action, V)
        print("{:<10} {:<12} {:<12.4f}".format(state, action, q))

print("\nOPTIMAL POLICY")
for state in ['S0', 'S1']:
    print(state, ":", policy[state])

print("G : Terminal")

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    max_q = max(calculate_Q(state, action, V) for action in actions)
    print(state, "V*(s) =", round(V[state], 4),
          "Max Q*(s,a) =", round(max_q, 4))

print("G: V*(G) = 0")

print("\nOPTIMAL PATH: S0 -> S1 -> G")
```

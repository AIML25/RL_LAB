```python

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.000001

positions = {
    'S0': (0, 0),
    'S1': (1, 0),
    'G': (1, 1)
}

wall = (0, 1)

moves = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

def transition(state, action):
    if state == 'G':
        return 'G', 0

    r, c = positions[state]
    dr, dc = moves[action]
    new_position = (r + dr, c + dc)

    if (new_position == wall or
        new_position[0] < 0 or new_position[0] > 1 or
        new_position[1] < 0 or new_position[1] > 1):
        return state, -1

    if new_position == positions['G']:
        return 'G', 100

    for s, pos in positions.items():
        if pos == new_position:
            return s, -1

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

iteration = 0

print("TRANSITION INFORMATION")
print("----------------------")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transition(state, action)
        print(state, action, "->", next_state, "Reward:", reward)

print("\nITERATION-WISE STATE VALUES")
print("---------------------------")

while True:
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state, reward = transition(state, action)
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])

    V = new_V
    iteration += 1

    print("Iteration", iteration,
          "S0 =", round(V['S0'], 4),
          "S1 =", round(V['S1'], 4))

    if delta < theta:
        break

print("\nFINAL OPTIMAL STATE VALUES")

for state in states:
    print(state, ":", round(V[state], 4))

print("\nFINAL ACTION VALUE TABLE")

print("{:<10} {:<12} {:<12} {:<12}".format(
    "State", "Action", "Next State", "Q Value"
))

policy = {}

for state in ['S0', 'S1']:
    q_values = {}

    for action in actions:
        next_state, reward = transition(state, action)
        q = reward + gamma * V[next_state]
        q_values[action] = q

        print("{:<10} {:<12} {:<12} {:<12.4f}".format(
            state, action, next_state, q
        ))

    policy[state] = max(q_values, key=q_values.get)

print("\nOPTIMAL POLICY")

for state in ['S0', 'S1']:
    print(state, ":", policy[state])

print("G : Terminal")

print("\nOPTIMAL PATH: S0 -> S1 -> G")
```

```python
n = int(input("Enter number of machines: "))

names = []
q_values = []

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    total = 0

    for reward in rewards:
        total = total + reward

    q = total / len(rewards)

    names.append(name)
    q_values.append(q)

print("\n--- Estimated Values ---")

for i in range(n):
    print("Machine", names[i])
    print("Q(" + names[i] + ") =", round(q_values[i], 2))

highest = 0

for i in range(1, n):
    if q_values[i] > q_values[highest]:
        highest = i

print("\nGreedy Selected Machine:", names[highest])
```

```python
n = int(input("Enter number of machines: "))

names = []
rewards_list = []
Q = []
N = []


for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    names.append(name)
    rewards_list.append(rewards)
    Q.append(0)
    N.append(0)



print("\n--- Incremental Updates ---")

for i in range(n):
    print("\nMachine", names[i])

    for reward in rewards_list[i]:

        N[i] = N[i] + 1

        # Incremental update formula
        Q[i] = Q[i] + (reward - Q[i]) / N[i]

        print("Reward =", reward)
        print("N =", N[i])
        print("Q =", round(Q[i], 2))



print("\n--- Final Estimated Values ---")

for i in range(n):
    print("Machine", names[i])
    print("Rewards:", rewards_list[i])
    print("N =", N[i])
    print("Q =", round(Q[i], 2))



highest = 0

for i in range(1, n):
    if Q[i] > Q[highest]:
        highest = i

print("\nGreedy Selected Machine:", names[highest])
```

```python

gamma = 0.9
V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

V_S0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V_S1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("ASSIGNMENT 1: ALGEBRAIC SOLUTION")
print("-" * 45)

print(f"V_pi(S0)       = {V_S0:.2f}")
print(f"V_pi(S1)       = {V_S1:.2f}")

print("\nAction Values:")
print(f"Q_pi(S0, Down)  = {Q_S0_Down:.2f}")
print(f"Q_pi(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q_pi(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q_pi(S1, Up)    = {Q_S1_Up:.2f}")

print("\nBellman Equation Verification:")
print(f"V_pi(S0) check = {V_S0_check:.2f}")
print(f"V_pi(S1) check = {V_S1_check:.2f}")

print("\nVerification successful:",
      abs(V_S0 - V_S0_check) < 1e-9 and
      abs(V_S1 - V_S1_check) < 1e-9)
```

```python

gamma = 0.9
theta = 0.000001

V_S0 = 0.0
V_S1 = 0.0

iterations = 0

while True:
    old_V_S0 = V_S0
    old_V_S1 = V_S1

    V_S0 = (
        0.1 * (-1 + gamma * 0) +
        0.9 * (-1 + gamma * old_V_S1)
    )

    V_S1 = (
        0.9 * (100 + gamma * 0) +
        0.1 * (-1 + gamma * old_V_S0)
    )

    delta = max(
        abs(V_S0 - old_V_S0),
        abs(V_S1 - old_V_S1)
    )

    iterations += 1

    if delta < theta:
        break

Q_S0_Down = (
    0.1 * (-1 + gamma * 0) +
    0.9 * (-1 + gamma * V_S1)
)

Q_S0_Right = -1 + gamma * 0

Q_S1_Right = (
    0.9 * (100 + gamma * 0) +
    0.1 * (-1 + gamma * V_S0)
)

Q_S1_Up = -1 + gamma * V_S0

print("ASSIGNMENT 2: ITERATIVE POLICY EVALUATION")
print("-" * 50)

print(f"Number of iterations = {iterations}")
print(f"V_pi(S0)             = {V_S0:.6f}")
print(f"V_pi(S1)             = {V_S1:.6f}")

print("\nAction Values:")
print(f"Q_pi(S0, Down)       = {Q_S0_Down:.6f}")
print(f"Q_pi(S0, Right)      = {Q_S0_Right:.6f}")
print(f"Q_pi(S1, Right)      = {Q_S1_Right:.6f}")
print(f"Q_pi(S1, Up)         = {Q_S1_Up:.6f}")

print("\nPolicy Consistency Checks:")
print(f"V(S0) = Q(S0, Down): {abs(V_S0 - Q_S0_Down) < theta}")
print(f"V(S1) = Q(S1, Right): {abs(V_S1 - Q_S1_Right) < theta}")
```
